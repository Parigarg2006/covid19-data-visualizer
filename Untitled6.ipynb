```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(0)
dates = pd.date_range('2020-03-01', periods=180, freq='D')
cases_daily = np.abs(np.random.normal(5000, 2000, 180) * np.sin(np.linspace(0,np.pi,180)))
deaths_daily = cases_daily * np.random.uniform(0.01, 0.03, 180)
recovered_daily = cases_daily * np.random.uniform(0.7, 0.9, 180)

df = pd.DataFrame({'date': dates, 'cases': cases_daily,
                   'deaths': deaths_daily, 'recovered': recovered_daily})
df['active'] = df['cases'].cumsum() - df['deaths'].cumsum() - df['recovered'].cumsum()

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('COVID-19 Data Analysis', fontsize=16, fontweight='bold')

axes[0,0].fill_between(df['date'], df['cases'], alpha=0.6, color='red')
axes[0,0].set_title('Daily New Cases')
axes[0,0].set_ylabel('Cases')

axes[0,1].plot(df['date'], df['cases'].cumsum(), color='orange', label='Total Cases')
axes[0,1].plot(df['date'], df['recovered'].cumsum(), color='green', label='Recovered')
axes[0,1].plot(df['date'], df['deaths'].cumsum(), color='black', label='Deaths')
axes[0,1].set_title('Cumulative Stats')
axes[0,1].legend()

weekly = df.resample('W', on='date')['cases'].sum()
axes[1,0].bar(weekly.index, weekly.values, color='steelblue', width=5)
axes[1,0].set_title('Weekly Cases')

recovery_rate = (df['recovered'].cumsum() / df['cases'].cumsum() * 100).fillna(0)
axes[1,1].plot(df['date'], recovery_rate, color='green')
axes[1,1].axhline(y=80, color='red', linestyle='--', label='80% target')
axes[1,1].set_title('Recovery Rate (%)')
axes[1,1].legend()

plt.tight_layout()
plt.savefig('covid_analysis.png', dpi=150)
plt.show()
```
